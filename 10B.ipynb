```python
import random
import sys
import time

# Increase recursion limit to handle worst-case scenarios safely
sys.setrecursionlimit(30000)

class QuickSortComparison:
    def __init__(self):
        self.comparisons = 0

    def partition(self, arr, low, high, pivot_index):
        """Standard Lomuto partition using a specified pivot index."""
        # Move pivot to the end
        arr[pivot_index], arr[high] = arr[high], arr[pivot_index]
        pivot = arr[high]

        i = low - 1
        for j in range(low, high):
            self.comparisons += 1
            if arr[j] <= pivot:
                i += 1
                arr[i], arr[j] = arr[j], arr[i]

        arr[i + 1], arr[high] = arr[high], arr[i + 1]
        return i + 1

    # --- 1. Deterministic Quick Sort (Pivot = Last Element) ---
    def deterministic_partition(self, arr, low, high):
        return self.partition(arr, low, high, high)

    def _deterministic_sort(self, arr, low, high):
        if low < high:
            pi = self.deterministic_partition(arr, low, high)
            self._deterministic_sort(arr, low, pi - 1)
            self._deterministic_sort(arr, pi + 1, high)

    def sort_deterministic(self, arr):
        self.comparisons = 0
        start = time.perf_counter()
        self._deterministic_sort(arr, 0, len(arr) - 1)
        end = time.perf_counter()
        return end - start, self.comparisons

    # --- 2. Randomized Quick Sort (Pivot = Random Element) ---
    def randomized_partition(self, arr, low, high):
        rand_index = random.randint(low, high)
        return self.partition(arr, low, high, rand_index)

    def _randomized_sort(self, arr, low, high):
        if low < high:
            pi = self.randomized_partition(arr, low, high)
            self._randomized_sort(arr, low, pi - 1)
            self._randomized_sort(arr, pi + 1, high)

    def sort_randomized(self, arr):
        self.comparisons = 0
        start = time.perf_counter()
        self._randomized_sort(arr, 0, len(arr) - 1)
        end = time.perf_counter()
        return end - start, self.comparisons

    # --- 3. Median-of-Three Quick Sort (Pivot = Median of Low, Mid, High) ---
    def median_of_three_partition(self, arr, low, high):
        mid = (low + high) // 2

        # Find median of arr[low], arr[mid], arr[high] and choose its index as pivot
        a = arr[low]
        b = arr[mid]
        c = arr[high]

        if (a - b) * (c - a) >= 0:
            pivot_index = low
        elif (b - a) * (c - b) >= 0:
            pivot_index = mid
        else:
            pivot_index = high

        return self.partition(arr, low, high, pivot_index)

    def _median_of_three_sort(self, arr, low, high):
        if low < high:
            pi = self.median_of_three_partition(arr, low, high)
            self._median_of_three_sort(arr, low, pi - 1)
            self._median_of_three_sort(arr, pi + 1, high)

    def sort_median_of_three(self, arr):
        self.comparisons = 0
        start = time.perf_counter()
        self._median_of_three_sort(arr, 0, len(arr) - 1)
        end = time.perf_counter()
        return end - start, self.comparisons

def main():
    N = 10000
    print(f"--- Quick Sort Pivot Comparison (Array Size N = {N}) ---")
    print(f"{'Configuration':<16} | {'Strategy':<18} | {'Comparisons':<12} | {'Time (s)'}")
    print("-" * 65)

    base_rand = [random.randint(1, N) for _ in range(N)]
    configs = {
        "Random": base_rand,
        "Sorted": sorted(base_rand),
        "Reverse Sorted": sorted(base_rand, reverse=True)
    }

    for config_name, arr in configs.items():
        # Test Deterministic
        qs1 = QuickSortComparison()
        t_det, c_det = qs1.sort_deterministic(arr.copy())

        # Test Randomized
        qs2 = QuickSortComparison()
        t_rand, c_rand = qs2.sort_randomized(arr.copy())

        # Test Median-of-Three
        qs3 = QuickSortComparison()
        t_med, c_med = qs3.sort_median_of_three(arr.copy())

        print(f"{config_name:<16} | {'Deterministic':<18} | {c_det:<12} | {t_det:.6f}")
        print(f"{'':<16} | {'Randomized':<18} | {c_rand:<12} | {t_rand:.6f}")
        print(f"{'':<16} | {'Median-of-Three':<18} | {c_med:<12} | {t_med:.6f}")
        print("-" * 65)

if __name__ == "__main__":
    main()
```
