```python
import sys

def matrix_chain_order(p):
    """
    Computes the minimum number of scalar multiplications needed to multiply
    a chain of matrices and records the optimal split points.

    p: Array of dimensions where matrix Ai has dimension p[i-1] x p[i]
    """
    n = len(p) - 1

    # m[i][j] will store the minimum number of scalar multiplications needed
    # to compute the matrix A[i]A[i+1]...A[j]
    m = [[0 for _ in range(n + 1)] for _ in range(n + 1)]

    # s[i][j] will store the index of the optimal split point
    s = [[0 for _ in range(n + 1)] for _ in range(n + 1)]

    # l is the chain length
    for l in range(2, n + 1):
        for i in range(1, n - l + 2):
            j = i + l - 1
            m[i][j] = sys.maxsize

            # Try all possible split points k between i and j-1
            for k in range(i, j):
                # Cost = cost of left sub-chain + cost of right sub-chain
                # + cost of multiplying the two resulting matrices
                q = m[i][k] + m[k + 1][j] + p[i - 1] * p[k] * p[j]

                if q < m[i][j]:
                    m[i][j] = q
                    s[i][j] = k

    return m, s

def print_optimal_parens(s, i, j):
    """
    Recursively constructs the string representing the optimal parenthesization.
    """
    if i == j:
        return f"A{i}"
    else:
        left = print_optimal_parens(s, i, s[i][j])
        right = print_optimal_parens(s, s[i][j] + 1, j)
        return f"({left}{right})"

if __name__ == '__main__':
    # Given dimensions: A1(10x30), A2(30x5), A3(5x60), A4(60x10)
    # The array p represents these dimensions: p[0]=10, p[1]=30, p[2]=5, p[3]=60, p[4]=10
    p = [10, 30, 5, 60, 10]
    n = len(p) - 1

    # Calculate the m and s tables
    m, s = matrix_chain_order(p)

    # Retrieve optimal order string and minimum cost
    optimal_order = print_optimal_parens(s, 1, n)
    min_multiplications = m[1][n]

    print("--- Matrix Chain Multiplication (Dynamic Programming) ---")
    print(f"Matrices: A1(10x30), A2(30x5), A3(5x60), A4(60x10)")
    print("-" * 55)
    print(f"Optimal Parenthesization:       {optimal_order}")
    print(f"Minimum Scalar Multiplications: {min_multiplications}")
```
