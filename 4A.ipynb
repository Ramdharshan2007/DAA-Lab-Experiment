```python
import heapq
import sys

def dijkstra_directed(graph, source):
    """
    Computes shortest paths from a source to all other vertices in a directed graph.
    Returns the shortest distances and the predecessor map for path reconstruction.
    """
    n = len(graph)
    # Initialize distances to infinity and predecessors to None
    dist = [sys.maxsize] * n
    prev = [None] * n

    # Distance to source is always 0
    dist[source] = 0

    # Priority Queue stores tuples of (distance, vertex)
    pq = [(0, source)]
    visited = [False] * n

    while pq:
        # Extract vertex with minimum distance
        current_dist, u = heapq.heappop(pq)

        # Optimization: Skip if we've already found a shorter path to u
        # or if we've fully processed it
        if visited[u]:
            continue
        visited[u] = True

        # Relax adjacent edges
        for v, weight in graph.get(u, []):
            if not visited[v]:
                new_dist = current_dist + weight
                # If a shorter path is found
                if new_dist < dist[v]:
                    dist[v] = new_dist
                    prev[v] = u
                    heapq.heappush(pq, (new_dist, v))

    return dist, prev

def reconstruct_path(prev, source, target):
    """
    Reconstructs the path from source to target using the predecessor map.
    """
    path = []
    current = target

    # Trace back from target to source
    while current is not None:
        path.append(current)
        current = prev[current]

    path.reverse()

    # Verify a valid path exists
    if path and path[0] == source:
        return path
    return []

# --- Main Execution ---
if __name__ == '__main__':
    # 1. Define the Directed Weighted Graph
    # 6 vertices (0-5)
    # Format: adj_list[u] = [(v, weight), ...]
    directed_graph = {
        0: [(1, 5), (2, 2)],
        1: [(3, 4), (4, 2)],
        2: [(1, 8), (4, 7)],
        3: [(5, 3)],
        4: [(3, 6), (5, 1)],
        5: [] # Sink node
    }

    start_vertex = 0
    total_vertices = 6

    print(f"--- Dijkstra's Algorithm (Directed Graph) ---")
    print(f"Source Vertex: {start_vertex}\n")

    # 2. Run Algorithm
    distances, predecessors = dijkstra_directed(directed_graph, start_vertex)

    # 3. Display Results
    print(f"{'Destination':<12} | {'Distance':<10} | {'Shortest Path'}")
    print("-" * 55)

    for v in range(total_vertices):
        # Handle unreachable vertices
        if distances[v] == sys.maxsize:
            print(f"{v:<12} | {'Unreachable':<10} | None")
        else:
            # Reconstruct and format the path
            path = reconstruct_path(predecessors, start_vertex, v)
            path_str = " -> ".join(map(str, path))
            print(f"{v:<12} | {distances[v]:<10} | {path_str}")
```
