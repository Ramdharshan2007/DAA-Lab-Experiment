```python
import heapq

# --- Kruskal's Algorithm Components ---
class UnionFind:
    """Disjoint Set data structure with path compression and union by rank."""
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, i):
        if self.parent[i] == i:
            return i
        self.parent[i] = self.find(self.parent[i]) # Path compression
        return self.parent[i]

    def union(self, i, j):
        root_i = self.find(i)
        root_j = self.find(j)
        if root_i != root_j:
            # Union by rank
            if self.rank[root_i] < self.rank[root_j]:
                self.parent[root_i] = root_j
            elif self.rank[root_i] > self.rank[root_j]:
                self.parent[root_j] = root_i
            else:
                self.parent[root_j] = root_i
                self.rank[root_i] += 1
            return True
        return False

def kruskal_mst(vertices, edges):
    """Computes MST using Kruskal's Algorithm."""
    result = []
    edges = sorted(edges, key=lambda item: item[0]) # Sort edges by weight
    uf = UnionFind(vertices)
    total_weight = 0
    i, e = 0, 0

    while e < vertices - 1 and i < len(edges):
        w, u, v = edges[i]
        i += 1
        # If including this edge does not cause a cycle
        if uf.union(u, v):
            result.append((u, v, w))
            total_weight += w
            e += 1

    return result, total_weight

# --- Prim's Algorithm Components ---
def prim_mst(vertices, edges_list):
    """Computes MST using Prim's Algorithm."""
    # Convert edge list to adjacency list
    adj = {i: [] for i in range(vertices)}
    for w, u, v in edges_list:
        adj[u].append((w, v))
        adj[v].append((w, u))

    min_heap = [(0, 0, -1)] # (weight, current_vertex, parent_vertex)
    visited = set()
    total_weight = 0
    mst_edges = []

    while min_heap and len(visited) < vertices:
        weight, u, parent = heapq.heappop(min_heap)

        if u not in visited:
            visited.add(u)
            if parent != -1: # Ignore the arbitrary parent of the starting node
                mst_edges.append((parent, u, weight))
                total_weight += weight

            for next_weight, v in adj[u]:
                if v not in visited:
                    heapq.heappush(min_heap, (next_weight, v, u))

    return mst_edges, total_weight

# --- Main Execution ---
if __name__ == '__main__':
    V = 7 # 7 Vertices (0 to 6)

    # 9 Edges: (weight, u, v)
    E = [
        (2, 0, 1),
        (7, 0, 3),
        (3, 1, 2),
        (8, 1, 4),
        (1, 2, 4),
        (5, 3, 4),
        (6, 3, 5),
        (4, 4, 6),
        (2, 5, 6)
    ]

    print("--- Kruskal's Algorithm ---")
    k_mst, k_weight = kruskal_mst(V, E)
    for u, v, w in k_mst:
        print(f"Edge: {u}-{v}, Weight: {w}")
    print(f"Total Weight (Kruskal): {k_weight}\n")

    print("--- Prim's Algorithm ---")
    p_mst, p_weight = prim_mst(V, E)
    for u, v, w in p_mst:
        print(f"Edge: {u}-{v}, Weight: {w}")
    print(f"Total Weight (Prim): {p_weight}\n")

    print("--- Verification ---")
    if k_weight == p_weight:
        print(f"Success: Both algorithms produced the exact same minimum weight of {k_weight}.")
    else:
        print("Error: Weights do not match.")
```
