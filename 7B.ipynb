```python
import time

def solve_n_queens_bitmask(n):
    """
    Solves the N-Queens problem using Bitmasking for O(1) safety checks.
    Returns the total number of solutions and the number of backtracks.
    """
    solutions_count = [0]
    backtrack_count = [0]

    # We use integers as bit arrays to track occupied spaces.
    # 0 means safe, 1 means occupied.
    cols = 0
    diag1 = 0  # Tracks main diagonals (row + col)
    diag2 = 0  # Tracks anti-diagonals (row - col + n - 1)

    def is_safe(row, col):
        """
        Checks if a position is safe in O(1) time using bitwise AND.
        If any of the bits at the specific column or diagonal index are 1, it's unsafe.
        """
        return not (
            (cols & (1 << col)) or
            (diag1 & (1 << (row + col))) or
            (diag2 & (1 << (row - col + n - 1)))
        )

    def backtrack(row):
        nonlocal cols, diag1, diag2

        # Base case: All queens are placed successfully
        if row == n:
            solutions_count[0] += 1
            return

        # Try placing a queen in each column of the current row
        for col in range(n):
            if is_safe(row, col):
                # PLACE QUEEN: Set the corresponding bits to 1
                cols |= (1 << col)
                diag1 |= (1 << (row + col))
                diag2 |= (1 << (row - col + n - 1))

                # Recurse to the next row
                backtrack(row + 1)

                # BACKTRACK: Remove the queen by clearing the bits to 0
                cols &= ~(1 << col)
                diag1 &= ~(1 << (row + col))
                diag2 &= ~(1 << (row - col + n - 1))

                # Increment backtrack counter when undoing a step
                backtrack_count[0] += 1

    backtrack(0)
    return solutions_count[0], backtrack_count[0]

if __name__ == '__main__':
    n = 12

    print(f"--- N-Queens Problem (Bitmask Optimization) ---")
    print(f"Solving for N = {n}...\n")

    start_time = time.perf_counter()
    total_solutions, total_backtracks = solve_n_queens_bitmask(n)
    end_time = time.perf_counter()

    print(f"{'Metric':<20} | {'Result'}")
    print("-" * 40)
    print(f"{'Total Solutions':<20} | {total_solutions}")
    print(f"{'Total Backtracks':<20} | {total_backtracks:,}")
    print(f"{'Execution Time':<20} | {end_time - start_time:.4f} seconds")
```
