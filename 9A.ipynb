```python
import math

# --- 1. Helper Functions ---
def theoretical_lower_bound(weights, capacity):
    """Calculates the absolute minimum number of bins required."""
    return math.ceil(sum(weights) / capacity)

# --- 2. Approximation Algorithms ---

def first_fit(weights, capacity):
    """
    First Fit (FF): Places the item in the first bin that has enough space.
    If no bin has space, opens a new bin.
    """
    bins = []
    for w in weights:
        placed = False
        for b in bins:
            # Using round to avoid floating-point arithmetic errors (e.g., 0.7 + 0.3 = 1.0000000000000002)
            if round(sum(b) + w, 4) <= capacity:
                b.append(w)
                placed = True
                break
        if not placed:
            bins.append([w])
    return bins

def best_fit(weights, capacity):
    """
    Best Fit (BF): Places the item in the tightest spot possible
    (the bin that will have the least space left over after placement).
    """
    bins = []
    for w in weights:
        best_bin_idx = -1
        min_space_left = float('inf')

        for i, b in enumerate(bins):
            space_left = round(capacity - sum(b), 4)
            # Check if it fits, and if it leaves a tighter remainder than previous best
            if space_left >= w and (space_left - w) < min_space_left:
                best_bin_idx = i
                min_space_left = space_left - w

        if best_bin_idx != -1:
            bins[best_bin_idx].append(w)
        else:
            bins.append([w])
    return bins

def first_fit_decreasing(weights, capacity):
    """
    First Fit Decreasing (FFD): Sorts the items in descending order first,
    then applies the First Fit strategy.
    """
    # Sort weights from largest to smallest
    sorted_weights = sorted(weights, reverse=True)
    return first_fit(sorted_weights, capacity)

# --- 3. Main Execution ---
if __name__ == '__main__':
    weights = [0.5, 0.7, 0.3, 0.9, 0.2, 0.6, 0.8, 0.4, 0.1, 0.5]
    capacity = 1.0

    print("--- 1D Bin Packing Problem ---")
    print(f"Item Weights: {weights}")
    print(f"Bin Capacity: {capacity}")

    # Calculate Lower Bound
    lb = theoretical_lower_bound(weights, capacity)
    print(f"\nTheoretical Lower Bound: {lb} bins")

    # Run First Fit
    ff_bins = first_fit(weights, capacity)
    print(f"\n[1] First Fit (FF) Used: {len(ff_bins)} bins")
    print(f"    Bin Contents: {ff_bins}")

    # Run Best Fit
    bf_bins = best_fit(weights, capacity)
    print(f"\n[2] Best Fit (BF) Used: {len(bf_bins)} bins")
    print(f"    Bin Contents: {bf_bins}")

    # Run First Fit Decreasing
    ffd_bins = first_fit_decreasing(weights, capacity)
    print(f"\n[3] First Fit Decreasing (FFD) Used: {len(ffd_bins)} bins")
    print(f"    Bin Contents: {ffd_bins}")

    # Analysis
    print("\n--- Accuracy Inference ---")
    print("The theoretical lower bound requires a minimum of 5 bins (Total Weight 5.0 / Capacity 1.0).")
    if len(ffd_bins) == lb:
        print("Success: First Fit Decreasing (FFD) successfully found the optimal arrangement!")
    print("By placing the largest items first, FFD prevents large items from being stranded in new bins at the end of the process, a common pitfall for standard FF and BF.")
```
