```python
def is_safe(board, row, col):
    """
    Checks if a queen can be safely placed at board[row][col].
    We only need to check the previous rows since we place queens row by row.
    """
    for prev_row in range(row):
        placed_col = board[prev_row]

        # Check if they are in the same column
        if placed_col == col:
            return False

        # Check if they are on the same diagonal
        if abs(prev_row - row) == abs(placed_col - col):
            return False

    return True

def solve_n_queens(n):
    """
    Solves the N-Queens problem and returns all valid board configurations.
    board array stores the column index of the queen for each row.
    """
    board = [-1] * n
    solutions = []
    backtrack_count = [0]

    def backtrack(row):
        # Base case: If all queens are placed, we found a solution
        if row == n:
            solutions.append(board[:])
            return

        # Try placing a queen in all columns of the current row
        for col in range(n):
            if is_safe(board, row, col):
                board[row] = col
                backtrack(row + 1)

                # Undo the placement (Backtrack) to explore other configurations
                board[row] = -1
                backtrack_count[0] += 1

    backtrack(0)
    return solutions, backtrack_count[0]

def display_board(solution, n):
    """
    Helper function to visually print the chessboard.
    'Q' represents a queen, '.' represents an empty square.
    """
    print(' +' + '---+' * n)
    for row in range(n):
        print(' |', end='')
        for col in range(n):
            if solution[row] == col:
                print(' Q |', end='')
            else:
                print(' . |', end='')
        print()
        print(' +' + '---+' * n)

if __name__ == '__main__':
    print("--- N-Queens Problem Solver ---")

    # Solve for N=4 (Show all solutions) and N=8 (Count only)
    for n in [4, 8]:
        print(f"\nSolving for {n}-Queens...")
        solutions, backtracks = solve_n_queens(n)

        print(f'Results for N={n}: {len(solutions)} total solutions found with {backtracks} backtracks.')

        if n == 4:
            print(f'\nVisualizing all solutions for {n}-Queens:')
            for i, sol in enumerate(solutions, 1):
                print(f'\nSolution {i}: {sol}')
                display_board(sol, n)
```
