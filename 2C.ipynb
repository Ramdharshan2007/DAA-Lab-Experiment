```python
def search_2d_pattern(text_matrix, pattern_matrix):
    """
    Searches for a 2D pattern matrix inside a 2D text matrix using the Naive approach.
    Tracks and returns the top-left coordinates of matches and total element comparisons.
    """
    t_rows = len(text_matrix)
    t_cols = len(text_matrix[0])
    p_rows = len(pattern_matrix)
    p_cols = len(pattern_matrix[0])

    matches = []
    comparisons = 0

    # Iterate through the text matrix
    # We only go up to (t_rows - p_rows) and (t_cols - p_cols) to prevent out of bounds
    for i in range(t_rows - p_rows + 1):
        for j in range(t_cols - p_cols + 1):
            match_found = True

            # Check the pattern against the current 3x3 sub-grid
            for r in range(p_rows):
                for c in range(p_cols):
                    comparisons += 1
                    # If any element mismatches, break early
                    if text_matrix[i + r][j + c] != pattern_matrix[r][c]:
                        match_found = False
                        break
                # If a mismatch was found in the row loop, break the outer pattern loop
                if not match_found:
                    break

            # If the loops completed without setting match_found to False, we have a match
            if match_found:
                matches.append((i, j))

    return matches, comparisons


def print_matrix(matrix, name):
    print(f"{name} Matrix:")
    for row in matrix:
        print(" ".join(map(str, row)))
    print()


# --- Main Execution ---
if __name__ == "__main__":
    # 1. Define a 10x10 text matrix
    text_matrix = [
        [1, 2, 3, 4, 5, 6, 7, 8, 9, 0],
        [0, 9, 8, 7, 6, 5, 4, 3, 2, 1],
        [1, 3, 5, 7, 9, 2, 4, 6, 8, 0],
        [2, 4, 6, 8, 0, 1, 3, 5, 7, 9],
        [1, 1, 1, 2, 2, 2, 3, 3, 3, 4],
        [5, 5, 5, 6, 6, 6, 7, 7, 7, 8],
        [9, 9, 9, 0, 0, 0, 1, 1, 1, 2],
        [3, 3, 3, 4, 4, 4, 5, 5, 5, 6],
        [1, 2, 3, 1, 2, 3, 1, 2, 3, 1],
        [4, 5, 6, 4, 5, 6, 4, 5, 6, 4]
    ]

    # 2. Define a 3x3 pattern matrix (present at top-left (4,3) and (7,3))
    pattern_matrix = [
        [2, 2, 2],
        [6, 6, 6],
        [0, 0, 0]
    ]

    # Display matrices
    print_matrix(text_matrix, "10x10 Text")
    print_matrix(pattern_matrix, "3x3 Pattern")

    # 3. Perform 2D Naive Pattern Matching
    print("-" * 40)
    print("Executing 2D Naive Pattern Matching...")
    print("-" * 40)

    matches, total_comparisons = search_2d_pattern(text_matrix, pattern_matrix)

    # 4. Display Results
    if matches:
        print(f"Pattern found at top-left indices (row, col): {matches}")
    else:
        print("Pattern not found in the text matrix.")

    print(f"Total element comparisons made: {total_comparisons}")
```
