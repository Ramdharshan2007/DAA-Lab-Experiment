```python
import time

def is_valid(x, y, board):
    """
    Checks if the given (x, y) coordinate is within the 8x8 board
    and hasn't been visited yet.
    """
    return 0 <= x < 8 and 0 <= y < 8 and board[x][y] == -1

def print_board(board):
    """
    Helper function to cleanly format and print the 8x8 chessboard.
    """
    print("+" + "----+" * 8)
    for row in board:
        print("|", end="")
        for cell in row:
            # Format numbers to always be 2 digits (e.g., 00, 05, 12)
            print(f" {str(cell).rjust(2, '0')} |", end="")
        print("\n+" + "----+" * 8)

def get_onward_degree(x, y, board, move_x, move_y):
    """
    Warnsdorff's Heuristic Helper:
    Counts how many valid, unvisited onward moves are available from (x, y).
    """
    count = 0
    for i in range(8):
        next_x = x + move_x[i]
        next_y = y + move_y[i]
        if is_valid(next_x, next_y, board):
            count += 1
    return count

def solve_kt_util(n, board, curr_x, curr_y, move_x, move_y, pos, backtracks):
    """
    Recursive backtracking function to find the Knight's Tour.
    """
    # Base case: If 64 moves are successfully made (0 to 63)
    if pos == n ** 2:
        return True

    # Gather all valid next moves
    next_moves = []
    for i in range(8):
        next_x = curr_x + move_x[i]
        next_y = curr_y + move_y[i]
        if is_valid(next_x, next_y, board):
            # Calculate the onward degree for Warnsdorff's heuristic
            degree = get_onward_degree(next_x, next_y, board, move_x, move_y)
            next_moves.append((degree, next_x, next_y))

    # Sort moves based on the lowest onward degree (Warnsdorff's Rule)
    # This massive optimization prevents the algorithm from getting stuck in dead ends
    next_moves.sort()

    # Try all valid moves in the optimized order
    for degree, next_x, next_y in next_moves:
        board[next_x][next_y] = pos

        # Recursively attempt to complete the tour
        if solve_kt_util(n, board, next_x, next_y, move_x, move_y, pos + 1, backtracks):
            return True

        # BACKTRACK: If the path fails, undo the move and increment backtrack counter
        board[next_x][next_y] = -1
        backtracks[0] += 1

    return False

def solve_knights_tour():
    """
    Initializes the chessboard and starts the backtracking algorithm.
    """
    n = 8
    # Initialize the board with -1 (indicating unvisited)
    board = [[-1 for _ in range(n)] for _ in range(n)]

    # All 8 possible moves for a Knight (L-shapes)
    move_x = [2, 1, -1, -2, -2, -1, 1, 2]
    move_y = [1, 2, 2, 1, -1, -2, -2, -1]

    # The Knight starts at the top-left corner (0,0)
    board[0][0] = 0
    backtracks = [0]

    print(f"--- Knight's Tour Problem (8x8) ---")
    print("Starting execution from position (0, 0)...\n")

    start_time = time.perf_counter()

    # Start the backtracking process at move position 1
    if not solve_kt_util(n, board, 0, 0, move_x, move_y, 1, backtracks):
        print("Solution does not exist for the given starting point.")
    else:
        end_time = time.perf_counter()
        print("Success! Hamiltonian Path Found:")
        print_board(board)
        print(f"\nTotal Backtracks: {backtracks[0]}")
        print(f"Execution Time: {end_time - start_time:.4f} seconds")

if __name__ == '__main__':
    # Increase recursion depth for safety, though 64 is well within defaults
    import sys
    sys.setrecursionlimit(2000)
    solve_knights_tour()
```
