```python
import random
import time

def partition(arr, low, high):
    """
    Standard Lomuto partition scheme.
    Chooses the last element as the pivot and places it in its correct
    sorted position, moving smaller elements to the left and larger to the right.
    """
    pivot = arr[high]
    i = low
    for j in range(low, high):
        if arr[j] <= pivot:
            arr[i], arr[j] = arr[j], arr[i]
            i += 1
    arr[i], arr[high] = arr[high], arr[i]
    return i

def quickselect(arr, low, high, k):
    """
    Finds the k-th smallest element in an unsorted array using Divide and Conquer.
    Note: k is 1-indexed (e.g., k=1 is the minimum element).
    """
    # Base case: If k is out of bounds
    if k > 0 and k <= high - low + 1:
        # Partition the array around a pivot
        pivot_index = partition(arr, low, high)

        # If the pivot itself is the k-th smallest element
        if pivot_index - low == k - 1:
            return arr[pivot_index]

        # If the k-th smallest is in the left sub-array
        if pivot_index - low > k - 1:
            return quickselect(arr, low, pivot_index - 1, k)

        # Else, the k-th smallest is in the right sub-array.
        # We adjust k to account for the elements we are discarding on the left.
        return quickselect(arr, pivot_index + 1, high, k - (pivot_index - low + 1))

    return None

def main():
    print("--- QuickSelect (Divide and Conquer) ---")

    # Generate an array of 1000 random integers
    n = 1000
    arr = [random.randint(1, 10000) for _ in range(n)]

    # We must pass copies of the array to quickselect because
    # the partition function modifies the array in-place.

    # 1. Find the Minimum (k = 1)
    start_time = time.perf_counter()
    min_val = quickselect(arr.copy(), 0, n - 1, 1)
    time_min = (time.perf_counter() - start_time) * 1000

    # 2. Find the Median (k = n // 2)
    start_time = time.perf_counter()
    median_val = quickselect(arr.copy(), 0, n - 1, n // 2)
    time_med = (time.perf_counter() - start_time) * 1000

    # 3. Find the Maximum (k = n)
    start_time = time.perf_counter()
    max_val = quickselect(arr.copy(), 0, n - 1, n)
    time_max = (time.perf_counter() - start_time) * 1000

    # Verification using Python's built-in sort
    sorted_arr = sorted(arr)
    expected_min = sorted_arr[0]
    expected_med = sorted_arr[(n // 2) - 1]
    expected_max = sorted_arr[-1]

    print(f"\nArray Size: {n}")
    print("-" * 50)
    print(f"{'Target':<10} | {'Calculated':<10} | {'Expected':<10} | {'Time (ms)':<10}")
    print("-" * 50)
    print(f"{'Minimum':<10} | {min_val:<10} | {expected_min:<10} | {time_min:.4f}")
    print(f"{'Median':<10} | {median_val:<10} | {expected_med:<10} | {time_med:.4f}")
    print(f"{'Maximum':<10} | {max_val:<10} | {expected_max:<10} | {time_max:.4f}")

if __name__ == '__main__':
    # Increase recursion depth just in case the random pivot choices are very poor
    import sys
    sys.setrecursionlimit(2000)
    main()
```
