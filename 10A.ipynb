```python
import random
import sys
import time

# Increase recursion limit for worst-case scenarios (sorted arrays)
sys.setrecursionlimit(20000)

class QuickSort:
    def __init__(self):
        self.comparisons = 0

    def deterministic_partition(self, arr, low, high):
        """Standard Lomuto partition scheme using the last element as pivot."""
        pivot = arr[high]
        i = low - 1
        for j in range(low, high):
            self.comparisons += 1
            if arr[j] <= pivot:
                i += 1
                arr[i], arr[j] = arr[j], arr[i]
        arr[i + 1], arr[high] = arr[high], arr[i + 1]
        return i + 1

    def randomized_partition(self, arr, low, high):
        """Chooses a random pivot, swaps it with the last element, then partitions."""
        rand_pivot = random.randint(low, high)
        arr[rand_pivot], arr[high] = arr[high], arr[rand_pivot]
        return self.deterministic_partition(arr, low, high)

    def _deterministic_quicksort(self, arr, low, high):
        if low < high:
            pi = self.deterministic_partition(arr, low, high)
            self._deterministic_quicksort(arr, low, pi - 1)
            self._deterministic_quicksort(arr, pi + 1, high)

    def _randomized_quicksort(self, arr, low, high):
        if low < high:
            pi = self.randomized_partition(arr, low, high)
            self._randomized_quicksort(arr, low, pi - 1)
            self._randomized_quicksort(arr, pi + 1, high)

    def sort_deterministic(self, arr):
        self.comparisons = 0
        start_time = time.perf_counter()
        self._deterministic_quicksort(arr, 0, len(arr) - 1)
        end_time = time.perf_counter()
        return end_time - start_time, self.comparisons

    def sort_randomized(self, arr):
        self.comparisons = 0
        start_time = time.perf_counter()
        self._randomized_quicksort(arr, 0, len(arr) - 1)
        end_time = time.perf_counter()
        return end_time - start_time, self.comparisons

def generate_arrays(n):
    """Generates four different configurations of arrays of size n."""
    base_arr = [random.randint(1, n) for _ in range(n)]
    return {
        "Random": base_arr.copy(),
        "Sorted": sorted(base_arr),
        "Reverse Sorted": sorted(base_arr, reverse=True),
        "All Identical": [42] * n
    }

def main():
    N = 10000
    arrays = generate_arrays(N)

    print(f"--- Quick Sort Performance Comparison (Array Size: {N}) ---")
    print(f"{'Configuration':<16} | {'Algorithm':<14} | {'Comparisons':<12} | {'Time (s)'}")
    print("-" * 65)

    for config, arr in arrays.items():
        # Test Deterministic
        qs_det = QuickSort()
        arr_copy1 = arr.copy()
        time_det, comp_det = qs_det.sort_deterministic(arr_copy1)

        # Test Randomized
        qs_rand = QuickSort()
        arr_copy2 = arr.copy()
        time_rand, comp_rand = qs_rand.sort_randomized(arr_copy2)

        print(f"{config:<16} | {'Deterministic':<14} | {comp_det:<12} | {time_det:.6f}")
        print(f"{'':<16} | {'Randomized':<14} | {comp_rand:<12} | {time_rand:.6f}")
        print("-" * 65)

if __name__ == "__main__":
    main()
```
