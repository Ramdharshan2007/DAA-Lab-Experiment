```python
import itertools
import time
import math

# --- 1. Brute-Force Algorithm ---
def tsp_brute_force(cost_matrix, start_city=0):
    """
    Evaluates all possible (n-1)! Hamiltonian cycles to find the absolute minimum cost.
    """
    n = len(cost_matrix)
    # Generate all permutations of cities to visit (excluding the start city)
    cities_to_visit = list(range(n))
    cities_to_visit.remove(start_city)

    min_path = None
    min_cost = float('inf')

    # Iterate through every possible permutation
    for perm in itertools.permutations(cities_to_visit):
        current_cost = 0
        current_city = start_city
        valid_path = True

        # Calculate cost for this specific permutation
        for next_city in perm:
            cost = cost_matrix[current_city][next_city]
            if cost == float('inf'):
                valid_path = False
                break
            current_cost += cost
            current_city = next_city

        # Add the cost to return to the starting city
        return_cost = cost_matrix[current_city][start_city]
        if valid_path and return_cost != float('inf'):
            current_cost += return_cost

            # Update minimums
            if current_cost < min_cost:
                min_cost = current_cost
                min_path = [start_city] + list(perm) + [start_city]

    return min_path, min_cost

# --- 2. Nearest-Neighbour Heuristic ---
def tsp_nearest_neighbor(cost_matrix, start_city=0):
    """
    Greedy approach: Always visits the closest unvisited city next.
    """
    n = len(cost_matrix)
    unvisited = set(range(n))
    unvisited.remove(start_city)

    path = [start_city]
    current_city = start_city
    total_cost = 0

    # While there are still cities left to visit
    while unvisited:
        nearest_city = None
        min_dist = float('inf')

        # Find the absolute closest unvisited city from the current position
        for next_city in unvisited:
            cost = cost_matrix[current_city][next_city]
            if cost < min_dist:
                min_dist = cost
                nearest_city = next_city

        # If no reachable city is found, a Hamiltonian cycle isn't possible from here
        if nearest_city is None:
            return [], float('inf')

        # Move to the nearest city
        path.append(nearest_city)
        total_cost += min_dist
        unvisited.remove(nearest_city)
        current_city = nearest_city

    # Return to the starting city
    return_cost = cost_matrix[current_city][start_city]
    if return_cost == float('inf'):
        return [], float('inf')

    path.append(start_city)
    total_cost += return_cost

    return path, total_cost

# --- Main Execution ---
if __name__ == '__main__':
    INF = float('inf')

    # 6-City Cost Matrix representing road distances (in km)
    # The matrix is symmetric in this specific example
    city_graph = [
        [INF, 12,  29,  22,  13,  24],
        [12,  INF, 19,  3,   25,  6 ],
        [29,  19,  INF, 21,  23,  28],
        [22,  3,   21,  INF, 4,   5 ],
        [13,  25,  23,  4,   INF, 16],
        [24,  6,   28,  5,   16,  INF]
    ]

    print("--- TSP: Brute Force vs Nearest Neighbour ---")
    print(f"Number of Cities: {len(city_graph)}")

    # 1. Run Brute Force
    start_bf = time.perf_counter()
    bf_path, bf_cost = tsp_brute_force(city_graph)
    time_bf = time.perf_counter() - start_bf

    # 2. Run Nearest Neighbour
    start_nn = time.perf_counter()
    nn_path, nn_cost = tsp_nearest_neighbor(city_graph)
    time_nn = time.perf_counter() - start_nn

    # 3. Format and Display Results
    print("\n[1] Brute Force (Exact Optimal)")
    print(f"    Path: {' -> '.join(map(str, bf_path))}")
    print(f"    Cost: {bf_cost} km")
    print(f"    Time: {time_bf:.6f} seconds")

    print("\n[2] Nearest Neighbour (Greedy Heuristic)")
    print(f"    Path: {' -> '.join(map(str, nn_path))}")
    print(f"    Cost: {nn_cost} km")
    print(f"    Time: {time_nn:.6f} seconds")

    # 4. Comparison Analysis
    print("\n--- Inference ---")
    if bf_cost == nn_cost:
        print("The Nearest Neighbour heuristic successfully found the exact optimal route!")
    else:
        diff = nn_cost - bf_cost
        pct = (diff / bf_cost) * 100
        print(f"The Nearest Neighbour heuristic path is longer by {diff} km (+{pct:.1f}% error).")
        print("This highlights that greedy, local choices do not always lead to a global optimum.")
```
