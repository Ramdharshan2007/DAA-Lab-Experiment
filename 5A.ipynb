```python
import random

def naive_min_max(arr):
    """
    Finds the minimum and maximum using a linear scan (Naive Approach).
    Returns a tuple of (min_val, max_val, comparison_count).
    """
    n = len(arr)
    if n == 0:
        return None, None, 0
    if n == 1:
        return arr[0], arr[0], 0

    comparisons = 0
    min_val = arr[0]
    max_val = arr[0]

    # Iterate through the array starting from the second element
    for i in range(1, n):
        comparisons += 1
        # Check if current element is greater than max
        if arr[i] > max_val:
            max_val = arr[i]
        else:
            # Only compare with min if it wasn't greater than max
            comparisons += 1
            if arr[i] < min_val:
                min_val = arr[i]

    return min_val, max_val, comparisons

def dac_min_max(arr, low, high):
    """
    Finds the minimum and maximum using Divide and Conquer.
    Returns a tuple of (min_val, max_val, comparison_count).
    """
    # Base Case 1: If there is only one element
    if low == high:
        return arr[low], arr[low], 0

    # Base Case 2: If there are exactly two elements
    if high == low + 1:
        if arr[low] > arr[high]:
            return arr[high], arr[low], 1
        else:
            return arr[low], arr[high], 1

    # Recursive Case: If there are more than 2 elements, divide the array
    mid = (low + high) // 2

    # Solve for the left half
    left_min, left_max, left_comp = dac_min_max(arr, low, mid)

    # Solve for the right half
    right_min, right_max, right_comp = dac_min_max(arr, mid + 1, high)

    # Combine (Merge) the solutions
    comparisons = left_comp + right_comp

    # Compare minimums
    comparisons += 1
    final_min = left_min if left_min < right_min else right_min

    # Compare maximums
    comparisons += 1
    final_max = left_max if left_max > right_max else right_max

    return final_min, final_max, comparisons

def performance_analysis():
    """
    Evaluates and prints the comparison counts for arrays of sizes
    10, 100, 1000, and 10000.
    """
    sizes = [10, 100, 1000, 10000]

    print(f"\n{'Array Size':>12} | {'Naive Comparisons':>20} | {'D&C Comparisons':>20}")
    print("-" * 60)

    for n in sizes:
        # Generate a random array of size n
        # We use random.sample to avoid duplicates, but random.choices works too
        arr = [random.randint(1, 100000) for _ in range(n)]

        # Execute Naive
        _, _, naive_comp = naive_min_max(arr)

        # Execute Divide and Conquer
        _, _, dac_comp = dac_min_max(arr, 0, len(arr) - 1)

        print(f"{n:>12} | {naive_comp:>20} | {dac_comp:>20}")

if __name__ == '__main__':
    print("--- Min-Max Finder: Divide & Conquer vs Naive ---")

    # Small Test Case
    test_arr = [1000, 11, 445, 1, 330, 3000]
    print(f"Sample Array: {test_arr}")

    d_min, d_max, d_comp = dac_min_max(test_arr, 0, len(test_arr) - 1)
    print(f"D&C Result -> Min: {d_min}, Max: {d_max}, Comparisons: {d_comp}")

    # Run the rigorous performance benchmarks
    performance_analysis()
```
