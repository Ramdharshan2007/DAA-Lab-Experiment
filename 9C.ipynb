```python
import random

def next_fit(weights, capacity):
    """
    Next Fit (NF): Places the item in the current bin.
    If it doesn't fit, the current bin is "closed" and a new bin is opened.
    It never looks back at previously closed bins.
    """
    bins = []
    if not weights:
        return bins

    current_bin = []
    current_sum = 0.0

    for w in weights:
        # Using round() to avoid floating point precision issues (e.g. 0.7 + 0.3 > 1.0)
        if round(current_sum + w, 4) <= capacity:
            current_bin.append(w)
            current_sum += w
        else:
            bins.append(current_bin)
            current_bin = [w]
            current_sum = w

    if current_bin:
        bins.append(current_bin)

    return bins

def first_fit(weights, capacity):
    """
    First Fit (FF): Places the item in the first bin that has enough space.
    Scans all existing bins before opening a new one.
    """
    bins = []
    for w in weights:
        placed = False
        for b in bins:
            if round(sum(b) + w, 4) <= capacity:
                b.append(w)
                placed = True
                break
        if not placed:
            bins.append([w])
    return bins

def best_fit(weights, capacity):
    """
    Best Fit (BF): Places the item in the tightest spot possible
    (the bin leaving the least remaining space).
    """
    bins = []
    for w in weights:
        best_bin_idx = -1
        min_space_left = float('inf')

        for i, b in enumerate(bins):
            space_left = round(capacity - sum(b), 4)
            if space_left >= w and (space_left - w) < min_space_left:
                best_bin_idx = i
                min_space_left = space_left - w

        if best_bin_idx != -1:
            bins[best_bin_idx].append(w)
        else:
            bins.append([w])
    return bins

def main():
    NUM_TRIALS = 100
    NUM_ITEMS = 50
    MIN_WEIGHT = 0.1
    MAX_WEIGHT = 0.9
    BIN_CAPACITY = 1.0

    total_nfd = 0
    total_ffd = 0
    total_bfd = 0

    print(f"--- Bin Packing Approximation Algorithms ---")
    print(f"Simulation: {NUM_TRIALS} Trials")
    print(f"Items per trial: {NUM_ITEMS} (Weights between {MIN_WEIGHT} and {MAX_WEIGHT})")
    print(f"Bin Capacity: {BIN_CAPACITY}\n")
    print("Running simulation...")

    random.seed(42) # Seed for reproducibility

    for _ in range(NUM_TRIALS):
        # Generate random weights
        items = [round(random.uniform(MIN_WEIGHT, MAX_WEIGHT), 4) for _ in range(NUM_ITEMS)]

        # Sort in descending order for the 'Decreasing' variants
        sorted_items = sorted(items, reverse=True)

        # Run algorithms and tally the number of bins used
        total_nfd += len(next_fit(sorted_items, BIN_CAPACITY))
        total_ffd += len(first_fit(sorted_items, BIN_CAPACITY))
        total_bfd += len(best_fit(sorted_items, BIN_CAPACITY))

    # Calculate Averages
    avg_nfd = total_nfd / NUM_TRIALS
    avg_ffd = total_ffd / NUM_TRIALS
    avg_bfd = total_bfd / NUM_TRIALS

    print("-" * 55)
    print(f"{'Algorithm':<30} | {'Average Bins Used':<20}")
    print("-" * 55)
    print(f"{'Next Fit Decreasing (NFD)':<30} | {avg_nfd:<20.2f}")
    print(f"{'First Fit Decreasing (FFD)':<30} | {avg_ffd:<20.2f}")
    print(f"{'Best Fit Decreasing (BFD)':<30} | {avg_bfd:<20.2f}")
    print("-" * 55)

    # Analysis
    print("\n--- Inference ---")
    print("NFD consistently uses the most bins because it 'closes' a bin forever as soon")
    print("as an item does not fit, missing out on filling small gaps with later, smaller items.")
    print("FFD and BFD check all previously opened bins, allowing them to pack much tighter.")

if __name__ == '__main__':
    main()
```
