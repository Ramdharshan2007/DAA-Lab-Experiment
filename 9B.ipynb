```python
import math

def theoretical_lower_bound(tasks, server_capacity):
    """Calculates the absolute minimum number of servers required."""
    return math.ceil(sum(tasks) / server_capacity)

def first_fit(tasks, capacity):
    """
    First Fit (FF): Places the task in the first server that has enough space.
    If no server has space, allocates a new server.
    """
    servers = []
    for t in tasks:
        placed = False
        for s in servers:
            # Using round to avoid floating-point arithmetic errors
            if round(sum(s) + t, 4) <= capacity:
                s.append(t)
                placed = True
                break
        if not placed:
            servers.append([t])
    return servers

def best_fit(tasks, capacity):
    """
    Best Fit (BF): Places the task in the tightest spot possible
    (the server that will have the least RAM left over after placement).
    """
    servers = []
    for t in tasks:
        best_server_idx = -1
        min_space_left = float('inf')

        for i, s in enumerate(servers):
            space_left = round(capacity - sum(s), 4)
            # Check if it fits, and if it leaves a tighter remainder than previous best
            if space_left >= t and (space_left - t) < min_space_left:
                best_server_idx = i
                min_space_left = space_left - t

        if best_server_idx != -1:
            servers[best_server_idx].append(t)
        else:
            servers.append([t])
    return servers

def first_fit_decreasing(tasks, capacity):
    """
    First Fit Decreasing (FFD): Sorts tasks in descending order first,
    then applies the First Fit strategy.
    """
    sorted_tasks = sorted(tasks, reverse=True)
    return first_fit(sorted_tasks, capacity)

def best_fit_decreasing(tasks, capacity):
    """
    Best Fit Decreasing (BFD): Sorts tasks in descending order first,
    then applies the Best Fit strategy.
    """
    sorted_tasks = sorted(tasks, reverse=True)
    return best_fit(sorted_tasks, capacity)

if __name__ == '__main__':
    # Provided RAM tasks (in GB)
    ram_tasks = [1.5, 2.0, 3.5, 0.5, 4.0, 1.0, 2.5, 3.0, 1.5, 2.0, 0.5, 4.0, 1.0, 3.5, 2.0]
    SERVER_CAPACITY = 8.0 # GB

    print("--- Cloud Server Resource Allocation (Bin Packing) ---")
    print(f"RAM Tasks (GB): {ram_tasks}")
    print(f"Total RAM Required: {sum(ram_tasks)} GB")
    print(f"Server Capacity: {SERVER_CAPACITY} GB")

    # Calculate Lower Bound
    lb = theoretical_lower_bound(ram_tasks, SERVER_CAPACITY)
    print(f"\nTheoretical Minimum Servers Required: {lb}")

    # Run FFD
    ffd_servers = first_fit_decreasing(ram_tasks, SERVER_CAPACITY)
    print(f"\n[1] First Fit Decreasing (FFD) Used: {len(ffd_servers)} servers")
    for i, s in enumerate(ffd_servers):
        print(f"    Server {i+1} [Used: {sum(s)}GB]: {s}")

    # Run BFD
    bfd_servers = best_fit_decreasing(ram_tasks, SERVER_CAPACITY)
    print(f"\n[2] Best Fit Decreasing (BFD) Used: {len(bfd_servers)} servers")
    for i, s in enumerate(bfd_servers):
        print(f"    Server {i+1} [Used: {sum(s)}GB]: {s}")

    # Analysis
    print("\n--- Inference ---")
    if len(ffd_servers) == lb and len(bfd_servers) == lb:
        print("Success: Both FFD and BFD achieved the optimal theoretical lower bound!")
    elif len(ffd_servers) == len(bfd_servers):
        print(f"Both FFD and BFD performed equally, using {len(ffd_servers)} servers.")
```
