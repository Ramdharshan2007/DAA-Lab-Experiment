```python
import sys
import time

def mcm_memoized(p, i, j, memo):
    """
    Top-Down Approach (Memoized Recursion)
    p: Array of dimensions
    i: Starting matrix index
    j: Ending matrix index
    memo: 2D array to cache computed results
    """
    # Base case: One matrix needs 0 multiplications
    if i == j:
        return 0

    # Return cached value if already computed
    if memo[i][j] != -1:
        return memo[i][j]

    min_cost = sys.maxsize

    # Try placing parenthesis at all possible positions between i and j-1
    for k in range(i, j):
        cost = (mcm_memoized(p, i, k, memo) +
                mcm_memoized(p, k + 1, j, memo) +
                p[i - 1] * p[k] * p[j])

        if cost < min_cost:
            min_cost = cost

    # Cache the computed minimum cost
    memo[i][j] = min_cost
    return min_cost

def mcm_bottom_up(p):
    """
    Bottom-Up Approach (Tabulation)
    p: Array of dimensions
    """
    n = len(p) - 1
    # m[i][j] will store the minimum number of scalar multiplications
    m = [[0 for _ in range(n + 1)] for _ in range(n + 1)]

    # l is the chain length
    for l in range(2, n + 1):
        for i in range(1, n - l + 2):
            j = i + l - 1
            m[i][j] = sys.maxsize

            for k in range(i, j):
                # Cost = cost of left + cost of right + multiplying the two
                cost = m[i][k] + m[k + 1][j] + p[i - 1] * p[k] * p[j]

                if cost < m[i][j]:
                    m[i][j] = cost

    # Return the minimum cost for multiplying matrices 1 through n
    return m[1][n]

if __name__ == '__main__':
    # Dimensions for a chain of 6 matrices (A1 to A6)
    # A1(30x35), A2(35x15), A3(15x5), A4(5x10), A5(10x20), A6(20x25)
    p = [30, 35, 15, 5, 10, 20, 25]
    n = len(p) - 1

    print("--- Matrix Chain Multiplication (6 Matrices) ---")
    print(f"Dimensions array (p): {p}\n")

    # --- 1. Top-Down Memoized ---
    # Initialize a memoization table with -1
    memo = [[-1 for _ in range(n + 1)] for _ in range(n + 1)]

    start_time = time.perf_counter()
    cost_memo = mcm_memoized(p, 1, n, memo)
    time_memo = time.perf_counter() - start_time

    # --- 2. Bottom-Up Tabulation ---
    start_time = time.perf_counter()
    cost_bu = mcm_bottom_up(p)
    time_bu = time.perf_counter() - start_time

    # --- Results & Verification ---
    print(f"{'Approach':<20} | {'Minimum Cost':<15} | {'Time (seconds)'}")
    print("-" * 55)
    print(f"{'Top-Down (Memoized)':<20} | {cost_memo:<15} | {time_memo:.6f}")
    print(f"{'Bottom-Up (Iterative)':<20} | {cost_bu:<15} | {time_bu:.6f}")

    print("\nVerification: ", end="")
    if cost_memo == cost_bu:
        print("SUCCESS - Both algorithms computed the exact same optimal cost!")
    else:
        print("ERROR - Cost mismatch detected.")
```
