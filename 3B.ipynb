```python
# Complete, well-commented, runnable code for this single file

class UnionFind:
    """Disjoint Set data structure to detect cycles, using dictionaries for string-based city names."""
    def __init__(self, vertices):
        # Initialize each vertex to be its own parent
        self.parent = {v: v for v in vertices}
        self.rank = {v: 0 for v in vertices}

    def find(self, item):
        # Path compression technique
        if self.parent[item] == item:
            return item
        self.parent[item] = self.find(self.parent[item])
        return self.parent[item]

    def union(self, x, y):
        # Union by rank technique
        root_x = self.find(x)
        root_y = self.find(y)

        if root_x != root_y:
            if self.rank[root_x] < self.rank[root_y]:
                self.parent[root_x] = root_y
            elif self.rank[root_x] > self.rank[root_y]:
                self.parent[root_y] = root_x
            else:
                self.parent[root_y] = root_x
                self.rank[root_x] += 1
            return True
        return False # Cycle detected

def kruskal_mst(vertices, edges):
    """Computes the Minimum Spanning Tree using Kruskal's Algorithm."""
    # 1. Sort edges by weight in ascending order
    edges = sorted(edges, key=lambda item: item[0])
    uf = UnionFind(vertices)

    mst = []
    total_cost = 0

    print("Step-by-step edge selection:")
    print("-" * 30)

    # 2. Iterate through sorted edges
    for weight, u, v in edges:
        # 3. If adding the edge doesn't form a cycle, include it
        if uf.union(u, v):
            mst.append((u, v, weight))
            total_cost += weight
            print(f"Accepted: {u} - {v} (Cost: {weight})")

            # Stop early if we have V - 1 edges
            if len(mst) == len(vertices) - 1:
                break
        else:
            print(f"Rejected: {u} - {v} (Cost: {weight}) - Forms a cycle")

    return mst, total_cost

# --- Main Execution ---
if __name__ == '__main__':
    # Define the 6 cities
    cities = ['A', 'B', 'C', 'D', 'E', 'F']

    # Define edges as (weight, city1, city2)
    connections = [
        (4, 'A', 'B'),
        (2, 'A', 'C'),
        (5, 'B', 'C'),
        (10, 'B', 'D'),
        (3, 'C', 'E'),
        (7, 'D', 'E'),
        (8, 'D', 'F'),
        (6, 'E', 'F')
    ]

    print(f"Total Cities: {len(cities)}")
    print(f"Total Connections: {len(connections)}\n")

    # Run algorithm
    mst_edges, min_cost = kruskal_mst(cities, connections)

    print("-" * 30)
    print("\nFinal Minimum Spanning Tree:")
    for u, v, w in mst_edges:
        print(f"{u} --- {v} : {w}")
    print(f"\nTotal Minimum Cost: {min_cost}")
```
