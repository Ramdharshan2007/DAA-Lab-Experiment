```python
import heapq

def dijkstra_fastest_route(graph, source, target):
    """
    Finds the fastest route from a source to a target node in a directed graph.
    Returns the total travel time and the sequence of nodes in the optimal path.
    """
    n = len(graph)

    dist = [float('inf')] * n
    prev = [None] * n
    dist[source] = 0

    pq = [(0, source)] # (accumulated_time, current_node)
    visited = set()

    while pq:
        current_time, u = heapq.heappop(pq)

        # Stop early if we have reached our target destination
        if u == target:
            break

        if u in visited:
            continue
        visited.add(u)

        # Check all outgoing one-way roads from the current node
        for v, travel_time in graph.get(u, []):
            if dist[u] + travel_time < dist[v]:
                dist[v] = dist[u] + travel_time
                prev[v] = u
                heapq.heappush(pq, (dist[v], v))

    path = []
    current = target
    while current is not None:
        path.append(current)
        current = prev[current]
    path.reverse()

    # Verify the path is valid and starts at our source
    if path and path[0] == source:
        return dist[target], path
    return float('inf'), []

if __name__ == '__main__':
    # Define the 5-node city road network (One-way roads)
    # Format: {u: [(v, time_in_minutes), ...]}
    city_graph = {
        0: [(1, 15), (2, 20)],       # Node 0: Airport
        1: [(2, 5), (3, 10)],        # Node 1: Suburb A
        2: [(3, 8), (4, 30)],        # Node 2: Highway Junction
        3: [(4, 10)],                # Node 3: Business District
        4: []                        # Node 4: Central Station (Destination)
    }

    # Map node IDs to readable names for better output
    node_names = {
        0: "Airport (Node 0)",
        1: "Suburb A (Node 1)",
        2: "Highway Junction (Node 2)",
        3: "Business District (Node 3)",
        4: "Central Station (Node 4)"
    }

    source_node = 0
    target_node = 4

    print("--- City Traffic Fastest Route Finder ---")
    print(f"Calculating route from {node_names[source_node]} to {node_names[target_node]}...\n")

    total_time, route = dijkstra_fastest_route(city_graph, source_node, target_node)

    if route:
        print(f"Fastest Travel Time: {total_time} minutes")
        print("Optimal Route:")
        # Format the path output cleanly
        for step, node in enumerate(route):
            if step < len(route) - 1:
                print(f"  {node_names[node]}\n   |\n   v")
            else:
                print(f"  {node_names[node]}")
    else:
        print("No valid route exists between the selected locations.")
```
